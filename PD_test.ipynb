```python
import numpy as np
import matplotlib.pyplot as plt

data = np.load("/home/sirous/Downloads/pd_run_0458_CH04/charge.npy")

print("Data shape:", data.shape)

# Angular coordinates
azimuthal_angles = np.linspace(0, 180, data.shape[1])  # 19 positions
polar_angles = np.linspace(25, 155, data.shape[0])     # 27 positions

plt.figure(figsize=(12, 7))

# Each line corresponds to one azimuthal angle
for i in range(data.shape[1]):
    plt.plot(
        polar_angles,
        data[:, i] * 1e12,
        label=f"Azimuthal = {azimuthal_angles[i]:.0f}°"
    )

plt.xlabel("Polar angle (degrees)")
plt.ylabel("Charge (pC)")
plt.title("Photodiode Charge vs Polar Angle run 458 old PD short cable")

plt.grid(True, alpha=0.3)
plt.legend(
    title="Azimuthal angle",
    bbox_to_anchor=(1.02, 1),
    loc="upper left"
)

plt.tight_layout()
plt.show()
```

```python
import numpy as np
import matplotlib.pyplot as plt

data = np.load("/home/sirous/Downloads/pd_run_0458_CH04/charge.npy")

print("Data shape:", data.shape)

# Convert array indices to angles
x_angles = np.linspace(0, 180, data.shape[1])
y_angles = np.linspace(25, 155, data.shape[0])

plt.figure(figsize=(12, 7))

plt.imshow(
    data * 1e12,
    aspect='auto',
    origin='lower',
    cmap='viridis',
    interpolation='nearest',
    extent=[
        x_angles[0] - 5,
        x_angles[-1] + 5,
        y_angles[0] - 2.5,
        y_angles[-1] + 2.5
    ]
)

plt.colorbar(label='Charge (pC)')
plt.xlabel('Azimuthal (degrees)')
plt.ylabel('Polar (degrees)')
plt.title('Photodiode Charge Heatmap run 458 old PD short cable')

plt.tight_layout()
plt.show()
```

```python
import numpy as np
import matplotlib.pyplot as plt

data = np.load("/home/sirous/Downloads/pd_run_0458_CH04/charge.npy")

print("Data shape:", data.shape)

# Angular coordinates
azimuthal_angles = np.linspace(0, 180, data.shape[1])  # X: 18 intervals
y_angles = np.linspace(25, 155, data.shape[0])         # Y: 26 intervals

plt.figure(figsize=(12, 7))

for i in range(data.shape[0]):
    plt.plot(
        azimuthal_angles,
        data[i] * 1e12,
        label=f"Y = {y_angles[i]:.0f}°"
    )

plt.xlabel("Azimuthal angle (degrees)")
plt.ylabel("Charge (pC)")
plt.title("Photodiode Charge vs Azimuthal Angle run 458 old PD short cable")

plt.grid(True, alpha=0.3)
plt.legend(title="Y angle", bbox_to_anchor=(1.02, 1), loc="upper left")

plt.tight_layout()
plt.show()
```
